# Giải thích các đặc trưng có trong bảng
### 1. Age: Tuổi của bệnh nhân
### 2. Sex: Giới tính
    Value 0: female
    Value 1: male
### 3. cp: Loại đau ngực
    Value 0: Đau thắt ngực
    Value 1: Đau thắt ngực không điển hình
    Value 2: Đau không phải do đau thắt ngực
    Value 3: Không có triệu chứng
### 4. trestbps: Huyết áp khi nghỉ ngơi 
### 5. chol: Nồng độ cholestoral trong huyết thanh (mg/dl)
### 6. fbs: Đường huyết lúc đói > 120 mg/dl
    Value 0: false
    Value 1: true
### 7. restecg: Kết quả điện tâm đồ lúc nghỉ ngơi
    Value 0: bình thườngS
    Value 1: có bất thường sóng ST-T (sóng T đảo ngược và/hoặc ST chênh lên hoặc chênh xuống > 0,05 mV)
    Value 2: cho thấy có khả năng hoặc chắc chắn bị phì đại thất trái theo tiêu chí của Estes.
### 8. thalach: Nhịp tim tối đa đạt đượcK
### 9. exang: Đau thắt ngực khi tập thể dục
    Value 0: No
    Value 1: Yes
### 10. oldpeak: Độ chênh xuống đoạn ST khi gắng sứcB
### 11. slope: Độ dốc của đoạn ST khi tập luyện ở mức cao nhất
### 12. ca: Số lượng mạch máu chính (0-3) được nhuộm màu bằng huỳnh quang
### 13. thal:
    Value 0: Lỗi giá trị (Giá trị naỳ không có ý nghĩa)
    Value 1: Khiếm khuyết cố định
    Value 2: Bình thường
    Value 3: Khiếm khuyết có thể hồi phụ
### 14. target: Mục tiêu 
    Value 0: Không mắc bệnh tim
    Value 1: Mắc bệnh tim

```python
#Convert data types 
features = ["sex", "cp", "fbs", "restecg", "exang", "slope", "thal"]
for i in features:
    data[i] = data[i].astype("category")
```
